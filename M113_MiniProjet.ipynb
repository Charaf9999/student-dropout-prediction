## 2. Découpage stratifié et encodage

Nous mélangeons les identifiants séparément dans chaque classe avec `default_rng(42)`, puis prenons 60 % pour train et 20 % pour validation, arrondis à l'entier inférieur. Le reste constitue le test. L'affectation sauvegardée fait référence pour les trois membres.

Les codes de filière, nationalité, profession et diplôme représentent des catégories. Le one-hot évite d'introduire une distance numérique arbitraire entre ces codes. Les variables binaires restent en 0/1. Les notes, quantités, âge, indicateurs économiques et ordre de candidature conservent leurs valeurs numériques. Le schéma est appris sur train. Une catégorie absente de train produit des zéros dans les colonnes one-hot de sa variable, et son occurrence est comptabilisée.

L'arbre utilise les valeurs avant standardisation. Les collègues ajusteront leur standardisation sur train. [Dictionnaire UCI](https://archive.ics.uci.edu/dataset/697/predict+students+dropout+and+academic+success)

````python
rng = np.random.default_rng(SEED)
split_ids = {name: [] for name in ["train", "val", "test"]}
for label in CLASSES:
    ids = rng.permutation(data.index[data["Target"].eq(label)].to_numpy())
    n_train, n_val = int(0.6 * len(ids)), int(0.2 * len(ids))
    split_ids["train"].extend(ids[:n_train].tolist())
    split_ids["val"].extend(ids[n_train:n_train + n_val].tolist())
    split_ids["test"].extend(ids[n_train + n_val:].tolist())

labels = data[["Target"]].copy()
labels["split"] = ""
for name, ids in split_ids.items():
    split_ids[name] = sorted(ids)
    labels.loc[ids, "split"] = name
labels = labels[["split", "Target"]]
assert labels["split"].ne("").all()
split_table = pd.crosstab(labels["split"], labels["Target"]).reindex(index=["train", "val", "test"], columns=CLASSES)
assert (split_table > 0).all().all()
display(split_table)

categorical_columns = [
    "Marital status", "Application mode", "Course", "Previous qualification",
    "Nacionality", "Mother's qualification", "Father's qualification",
    "Mother's occupation", "Father's occupation",
]
binary_columns = [
    "Daytime/evening attendance", "Displaced", "Educational special needs", "Debtor",
    "Tuition fees up to date", "Gender", "Scholarship holder", "International",
]
raw_features = data.drop(columns="Target")
numeric_columns = [column for column in raw_features if column not in categorical_columns]
assert set(categorical_columns + binary_columns).issubset(raw_features.columns)
categories = {
    column: sorted(raw_features.loc[split_ids["train"], column].unique().tolist())
    for column in categorical_columns
}

def encode_features(frame):
    encoded = {column: frame[column].astype(float) for column in numeric_columns}
    for column, values in categories.items():
        for value in values:
            encoded[f"{column}__{value}"] = frame[column].eq(value).astype(float)
    return pd.DataFrame(encoded, index=frame.index)

features_by_split = {name: encode_features(raw_features.loc[ids]) for name, ids in split_ids.items()}
target_by_split = {
    name: pd.Categorical(labels.loc[ids, "Target"], categories=CLASSES).codes.astype(int)
    for name, ids in split_ids.items()
}
mapping_rows = [
    {"feature": column, "source_column": column,
     "kind": "binary" if column in binary_columns else "numeric", "category": ""}
    for column in numeric_columns
]
for column, values in categories.items():
    mapping_rows.extend({"feature": f"{column}__{value}", "source_column": column,
                         "kind": "one_hot", "category": value} for value in values)
feature_mapping = pd.DataFrame(mapping_rows)
unknown_categories = {
    name: {column: int((~raw_features.loc[ids, column].isin(values)).sum())
           for column, values in categories.items()}
    for name, ids in split_ids.items()
}
print("Nombre de caractéristiques encodées", features_by_split["train"].shape[1])
print("Occurrences de catégories absentes de train", unknown_categories)
````

## 3. Arbre CART avec NumPy

Pour un nœud contenant des proportions de classes $p_c$, l'impureté vaut $G = 1-\sum_c p_c^2$. Nous choisissons la coupure qui maximise

$$\Delta G = G_{parent} - \frac{n_g}{n}G_g - \frac{n_d}{n}G_d.$$

Les seuils sont les milieux de valeurs distinctes consécutives. Les effectifs cumulés permettent d'évaluer les coupures sans recalculer chaque sous-tableau. Les ex æquo sont résolus par l'ordre des colonnes puis des seuils. Une feuille prédit sa classe majoritaire, avec l'ordre des classes comme départage.

L'importance d'une colonne additionne $(n_{nœud}/n_{train})\Delta G$ pour ses coupures. Les importances sont ensuite normalisées. Elles décrivent cet arbre et peuvent être partagées entre variables corrélées.

````python
class DecisionTreeNumpy:
    """Arbre binaire de classification, critère de Gini, sans scikit-learn."""

    def __init__(self, max_depth=None, min_samples_leaf=1, min_gain=1e-12):
        self.max_depth = max_depth
        self.min_samples_leaf = min_samples_leaf
        self.min_gain = min_gain

    def fit(self, X, y):
        values = np.asarray(X, dtype=float)
        target = np.asarray(y)
        if values.ndim != 2 or len(values) == 0 or values.shape[1] == 0:
            raise ValueError("X doit être une matrice non vide.")
        if target.ndim != 1 or len(target) != len(values) or not np.isfinite(values).all():
            raise ValueError("Vérifier les dimensions de y et les valeurs finies de X.")
        if self.min_samples_leaf < 1 or (self.max_depth is not None and self.max_depth < 0):
            raise ValueError("Profondeur >= 0 et taille minimale de feuille >= 1.")
        self.feature_names_in_ = list(X.columns) if isinstance(X, pd.DataFrame) else None
        self.classes_, encoded = np.unique(target, return_inverse=True)
        self.n_classes_ = len(self.classes_)
        self.n_features_in_ = values.shape[1]
        self.n_train_ = len(values)
        self.feature_importances_ = np.zeros(self.n_features_in_)
        self.node_count_ = 0
        self.depth_ = 0
        self.root_ = self._grow(values, encoded, depth=0)
        total = self.feature_importances_.sum()
        if total > 0:
            self.feature_importances_ /= total
        return self

    def _grow(self, X, y, depth):
        counts = np.bincount(y, minlength=self.n_classes_)
        impurity = 1.0 - np.sum((counts / len(y)) ** 2)
        node = {"prediction": int(np.argmax(counts)), "counts": counts.tolist(),
                "samples": len(y), "gini": float(impurity)}
        self.node_count_ += 1
        self.depth_ = max(self.depth_, depth)
        depth_limit = self.max_depth is not None and depth >= self.max_depth
        if impurity <= 1e-12 or depth_limit or len(y) < 2 * self.min_samples_leaf:
            return node

        best_gain = self.min_gain
        best_split = None
        for feature in range(X.shape[1]):
            order = np.argsort(X[:, feature], kind="stable")
            sorted_values = X[order, feature]
            positions = np.arange(1, len(y))
            valid = ((sorted_values[:-1] < sorted_values[1:])
                     & (positions >= self.min_samples_leaf)
                     & (len(y) - positions >= self.min_samples_leaf))
            positions = positions[valid]
            if len(positions) == 0:
                continue
            cumulative = np.cumsum(np.eye(self.n_classes_)[y[order]], axis=0)
            left_counts = cumulative[positions - 1]
            right_counts = counts - left_counts
            left_gini = 1.0 - np.sum((left_counts / positions[:, None]) ** 2, axis=1)
            right_sizes = len(y) - positions
            right_gini = 1.0 - np.sum((right_counts / right_sizes[:, None]) ** 2, axis=1)
            gains = impurity - (positions * left_gini + right_sizes * right_gini) / len(y)
            index = int(np.argmax(gains))
            gain = float(gains[index])
            if gain > best_gain + 1e-14:
                position = positions[index]
                lower, upper = sorted_values[position - 1], sorted_values[position]
                threshold = lower / 2.0 + upper / 2.0
                # Garantir une coupure valide même avec des flottants très proches.
                if threshold >= upper:
                    threshold = lower
                best_gain = gain
                best_split = (feature, float(threshold))

        if best_split is None:
            return node
        feature, threshold = best_split
        left_mask = X[:, feature] <= threshold
        self.feature_importances_[feature] += len(y) / self.n_train_ * best_gain
        node.update({"feature": feature, "threshold": threshold, "gain": best_gain,
                     "left": self._grow(X[left_mask], y[left_mask], depth + 1),
                     "right": self._grow(X[~left_mask], y[~left_mask], depth + 1)})
        return node

    def predict(self, X):
        if not hasattr(self, "root_"):
            raise ValueError("Appeler fit avant predict.")
        if isinstance(X, pd.DataFrame) and self.feature_names_in_ is not None:
            if list(X.columns) != self.feature_names_in_:
                raise ValueError("Les colonnes et leur ordre doivent correspondre à train.")
        values = np.asarray(X, dtype=float)
        if values.ndim != 2 or values.shape[1] != self.n_features_in_ or not np.isfinite(values).all():
            raise ValueError("Dimensions ou valeurs de X incorrectes.")
        predictions = []
        for row in values:
            node = self.root_
            while "feature" in node:
                node = node["left"] if row[node["feature"]] <= node["threshold"] else node["right"]
            predictions.append(node["prediction"])
        return self.classes_[np.asarray(predictions, dtype=int)]


def classification_metrics(y_true, y_pred, n_classes=3):
    matrix = np.zeros((n_classes, n_classes), dtype=int)
    np.add.at(matrix, (np.asarray(y_true, dtype=int), np.asarray(y_pred, dtype=int)), 1)
    true_positive = np.diag(matrix)
    denominator = matrix.sum(axis=0) + matrix.sum(axis=1)
    f1 = np.divide(2 * true_positive, denominator, out=np.zeros(n_classes), where=denominator > 0)
    return {"accuracy": float(true_positive.sum() / matrix.sum()),
            "f1_macro": float(f1.mean()), "f1_per_class": f1, "confusion_matrix": matrix}
````

## 4. Vérifications sur des exemples contrôlés

Ces tests vérifient une séparation connue, un problème à trois classes, un nœud pur, une colonne constante, la taille minimale des feuilles et le calcul du F1. Ils détectent des erreurs avant l'utilisation des données réelles.

````python
small_X = np.array([[0.], [1.], [2.], [3.]])
small_y = np.array([0, 0, 1, 1])
small_tree = DecisionTreeNumpy().fit(small_X, small_y)
assert small_tree.root_["threshold"] == 1.5
assert np.array_equal(small_tree.predict(small_X), small_y)
assert np.allclose(small_tree.feature_importances_, [1.0])
assert np.isclose(small_tree.root_["gain"], 0.5)

multi_X = np.arange(6).reshape(-1, 1)
multi_y = np.array([0, 0, 1, 1, 2, 2])
multi_tree = DecisionTreeNumpy().fit(multi_X, multi_y)
assert np.array_equal(multi_tree.predict(multi_X), multi_y)
pure_tree = DecisionTreeNumpy().fit(small_X, np.zeros(4, dtype=int))
assert pure_tree.node_count_ == 1 and pure_tree.feature_importances_.sum() == 0
constant_tree = DecisionTreeNumpy().fit(np.ones((4, 1)), small_y)
assert constant_tree.node_count_ == 1
assert np.array_equal(constant_tree.predict(np.ones((2, 1))), [0, 0])
limited_tree = DecisionTreeNumpy(min_samples_leaf=3).fit(small_X, small_y)
assert limited_tree.node_count_ == 1
assert np.isclose(classification_metrics([0, 1, 2], [0, 0, 2])["f1_macro"], 5 / 9)

def leaf_sizes(node):
    if "feature" not in node:
        return [node["samples"]]
    return leaf_sizes(node["left"]) + leaf_sizes(node["right"])

leaf_tree = DecisionTreeNumpy(min_samples_leaf=2).fit(multi_X, multi_y)
assert min(leaf_sizes(leaf_tree.root_)) >= 2
print("Vérifications de l'arbre et des métriques réussies.")
````

## 5. Comparaison et choix sur validation

Le modèle de base peut se développer jusqu'à des feuilles de taille 1. L'amélioration étudiée est le pré-élagage, en limitant la profondeur et la taille minimale des feuilles. Nous comparons neuf réglages. Le F1 macro donne le même poids à chaque classe.

La sélection du modèle utilise le meilleur F1 macro de validation. En cas d'égalité, nous retenons le plus petit arbre, puis le premier réglage essayé. Une seule partition de validation fournit une estimation qui dépend du découpage.

````python
tree_X_train = features_by_split["train"]
tree_X_val = features_by_split["val"]
tree_y_train = target_by_split["train"]
tree_y_val = target_by_split["val"]
configurations = [(None, 1)] + [(depth, leaf) for depth in [3, 5, 8] for leaf in [5, 10, 20]]
tree_models = []
comparison_rows = []
for depth, leaf in configurations:
    tree_model = DecisionTreeNumpy(max_depth=depth, min_samples_leaf=leaf)
    started = perf_counter()
    tree_model.fit(tree_X_train, tree_y_train)
    elapsed = perf_counter() - started
    train_scores = classification_metrics(tree_y_train, tree_model.predict(tree_X_train))
    val_scores = classification_metrics(tree_y_val, tree_model.predict(tree_X_val))
    comparison_rows.append({
        "model_id": len(tree_models), "variant": "base" if depth is None else "pre_elagage",
        "max_depth": "illimitee" if depth is None else str(depth), "min_samples_leaf": leaf,
        "actual_depth": tree_model.depth_, "nodes": tree_model.node_count_,
        "train_f1_macro": train_scores["f1_macro"], "val_f1_macro": val_scores["f1_macro"],
        "train_accuracy": train_scores["accuracy"], "val_accuracy": val_scores["accuracy"],
        "training_seconds": elapsed,
    })
    tree_models.append(tree_model)

comparison = pd.DataFrame(comparison_rows)
best_index = int(comparison.sort_values(["val_f1_macro", "nodes", "model_id"],
                                      ascending=[False, True, True]).iloc[0]["model_id"])
tree_best = tree_models[best_index]
comparison["retained"] = comparison["model_id"].eq(best_index)
comparison.to_csv(RESULTS / "tree_comparison.csv", index=False)
display(comparison.round(4))
print("Modèle retenu", best_index)
````

````python
fig, ax = plt.subplots(figsize=(11, 4))
positions = np.arange(len(comparison))
ax.plot(positions, comparison["train_f1_macro"], "o-", label="Entraînement", color="#3a8f85")
ax.plot(positions, comparison["val_f1_macro"], "o-", label="Validation", color="#d97757")
ax.set_xticks(positions, [f"{row.max_depth}/{row.min_samples_leaf}" for row in comparison.itertuples()], rotation=30)
ax.set(xlabel="Profondeur maximale / taille minimale de feuille", ylabel="F1 macro", ylim=(0, 1.03), title="Effet du pré-élagage")
ax.legend()
fig.tight_layout()
fig.savefig(FIGURES / "comparaison_arbres.png", bbox_inches="tight")
plt.show()
````

## 7. Exports communs et vérification du contrat

Le dossier `shared/` contient les identifiants, l'affectation des splits, le schéma et les données sélectionnées. Un identifiant de contenu permet de reconnaître une régénération identique. # Partie 2 — Classification : régression logistique et SVM

On charge les fichiers de `shared/` produits par la partie 1, on standardise avec un scaler appris sur train, puis on compare les modèles avec le F1 macro de validation. Le test ne sert qu'à l'évaluation finale.

### Imports

````python
import pandas as pd
import json
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.svm import SVC
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import classification_report
from sklearn.metrics import roc_curve, auc
````

### Loading data

````python
labels = pd.read_csv("shared/labels.csv", index_col="row_id")

X_train = pd.read_csv("shared/X_train.csv", index_col="row_id")
y_train = labels.loc[X_train.index, "Target"]
print(X_train.shape, y_train.shape)

X_val = pd.read_csv("shared/X_val.csv", index_col="row_id")
y_val = labels.loc[X_val.index, "Target"]

print(X_val.shape, y_val.shape)

X_test = pd.read_csv("shared/X_test.csv", index_col="row_id")
y_test = labels.loc[X_test.index, "Target"]

print(X_test.shape, y_test.shape)
````

### Data Checks

````python
#checking if the split column in the labels dataframe matches the training data
print(labels.loc[X_train.index, "split"].value_counts(dropna=False))

#checking if the order of the features in the data match the selected features
with open("shared/selected_features.json", encoding="utf-8") as file:
    features = json.load(file)["features"]

print(X_train.columns.tolist() == features)

print(X_val.columns.tolist() == features)
print(X_test.columns.tolist() == features)

#checking if each id appears only once across the splits
all_ids = X_train.index.append([X_val.index, X_test.index])
print(all_ids.is_unique)

#checking if the split column in the labels dataframe matches the validation and test data
print(labels.loc[X_val.index, "split"].value_counts(dropna=False))
print(labels.loc[X_test.index, "split"].value_counts(dropna=False))

#checking the distribution of the target variable in the data
print("Train\n", y_train.value_counts(dropna=False))
print("Validation\n", y_val.value_counts(dropna=False))
print("Test\n", y_test.value_counts(dropna=False))

#checking if all the splits contain valide values
for name, table in [("Train", X_train), ("Validation", X_val), ("Test", X_test)]:
    values = table.to_numpy()
    print(name, np.isfinite(values).all())
````

### Standarisation

````python
#fitting the scaler from the training set
scaler = StandardScaler()
scaler.fit(X_train)

#standardizing the data using the fitted scaler
X_train_scaled = scaler.transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
````

### Régression logistique

````python
#fitting the logistic regression model
logreg = LogisticRegression(max_iter=1000)
logreg.fit(X_train_scaled, y_train)

#predicting on the validation set
y_val_pred_logreg = logreg.predict(X_val_scaled)

#calculating the macro F1 score for the logistic regression model
logreg_val_f1 = f1_score(y_val, y_val_pred_logreg, average="macro")

print("Validation macro F1:", logreg_val_f1)
````

### SVM linéaire

````python
#fitting the SVM model with linear kernel
svm_linear = SVC(kernel="linear", C=1.0)
svm_linear.fit(X_train_scaled, y_train)

#predicting on the validation set
y_val_pred_linear = svm_linear.predict(X_val_scaled)

#calculating the macro F1 score for the linear SVM model
svm_linear_val_f1 = f1_score(y_val, y_val_pred_linear, average="macro")

print("Linear SVM validation macro F1:", svm_linear_val_f1)
````

### SVM à noyau RBF

````python
#fitting the SVM model with RBF kernel
svm_rbf = SVC(kernel="rbf", C=1.0, gamma="scale")
svm_rbf.fit(X_train_scaled, y_train)

#predicting on the validation set
y_val_pred_rbf = svm_rbf.predict(X_val_scaled)

#calculating the macro F1 score for the RBF SVM model
svm_rbf_val_f1 = f1_score(y_val, y_val_pred_rbf, average="macro")

print("RBF SVM validation macro F1:", svm_rbf_val_f1)
````

### Réglage de gamma

````python
#testing different values of gamma for the RBF kernel
rbf_gamma_scores = {"scale": svm_rbf_val_f1}

for gamma in [0.01, 1.0]:
    model = SVC(kernel="rbf", C=1.0, gamma=gamma)
    model.fit(X_train_scaled, y_train)

    predictions = model.predict(X_val_scaled)
    rbf_gamma_scores[gamma] = f1_score(y_val, predictions, average="macro")

print(rbf_gamma_scores)
````

### Réglage de C pour RBF SVM

````python
#testing different values of C for the RBF kernel with gamma=1.0
rbf_c_scores = {1.0: rbf_gamma_scores[1.0]}

for c_value in [0.1, 10.0]:
    model = SVC(kernel="rbf", C=c_value, gamma=1.0)
    model.fit(X_train_scaled, y_train)
    predictions = model.predict(X_val_scaled)
    rbf_c_scores[c_value] = f1_score(y_val, predictions, average="macro")

print(rbf_c_scores)
````

### Réglage de C pour SVM lineaire

````python
#testing different values of C for the linear kernel
linear_c_scores = {1.0: svm_linear_val_f1}

for c_value in [0.1, 10.0]:
    model = SVC(kernel="linear", C=c_value)
    model.fit(X_train_scaled, y_train)
    predictions = model.predict(X_val_scaled)
    linear_c_scores[c_value] = f1_score(y_val, predictions, average="macro")

print(linear_c_scores)
````

### Réglage de C de la régression logistique

````python
#testing different values of C for the logistic regression model
logreg_c_scores = {1.0: logreg_val_f1}

for c_value in [0.1, 10.0]:
    model = LogisticRegression(C=c_value, max_iter=1000)
    model.fit(X_train_scaled, y_train)
    predictions = model.predict(X_val_scaled)
    logreg_c_scores[c_value] = f1_score(y_val, predictions, average="macro")

print(logreg_c_scores)
````

### Pondération des classes pour regression logistique

````python
#testing the logistic regression model with class weights to handle class imbalance
logreg_balanced = LogisticRegression(
    C=10.0, max_iter=1000, class_weight="balanced"
)
logreg_balanced.fit(X_train_scaled, y_train)

predictions = logreg_balanced.predict(X_val_scaled)
logreg_balanced_val_f1 = f1_score(y_val, predictions, average="macro")

print("Unweighted:", logreg_c_scores[10.0]) #10 beause it was the best value of C in the previous test
print("Balanced:", logreg_balanced_val_f1)
````

### Pondération du SVM linéaire

````python
#testing the linear SVM model with class weights to handle class imbalance
svm_linear_balanced = SVC(kernel="linear", C=10.0, class_weight="balanced")
svm_linear_balanced.fit(X_train_scaled, y_train)

predictions = svm_linear_balanced.predict(X_val_scaled)
svm_linear_balanced_val_f1 = f1_score(y_val, predictions, average="macro")

print("Unweighted:", linear_c_scores[10.0])
print("Balanced:", svm_linear_balanced_val_f1)
````

````python
#testing the RBF SVM model with class weights to handle class imbalance
svm_rbf_balanced = SVC(
    kernel="rbf", C=1.0, gamma=1.0, class_weight="balanced"
)
svm_rbf_balanced.fit(X_train_scaled, y_train)

predictions = svm_rbf_balanced.predict(X_val_scaled)
svm_rbf_balanced_val_f1 = f1_score(y_val, predictions, average="macro")

print("Unweighted:", rbf_c_scores[1.0])
print("Balanced:", svm_rbf_balanced_val_f1)
````

## 3. Choix du nombre de groupes

Pour un étudiant $i$, $a(i)$ est la distance moyenne aux membres de son groupe et $b(i)$ la distance moyenne au groupe voisin le plus proche. Son coefficient de silhouette vaut

$$s(i) = \frac{b(i) - a(i)}{\max\{a(i), b(i)\}} \in [-1, 1].$$

Le score de silhouette est la moyenne des $s(i)$ (à maximiser). On teste k de 2 à 10 sur train, dans l'espace standardisé, avec `init="k-means++"`, `n_init=20` et `random_state=42`. On vérifie aussi que le nombre de groupes réellement formés est compris entre 2 et n − 1 (sinon la silhouette n'est pas définie). L'inertie (méthode du coude) et l'indice de Davies-Bouldin (à minimiser) servent de compléments. **Ce choix n'utilise pas `Target`.**

````python
def safe_silhouette(Z, cluster_labels):
    """Silhouette si 2 <= nombre de groupes formés <= n - 1, sinon NaN (cas dégénéré)."""
    formed = len(np.unique(cluster_labels))
    if 2 <= formed <= len(Z) - 1:
        return float(silhouette_score(Z, cluster_labels))
    return float("nan")


rows, km_models = [], {}
for k in K_RANGE:
    model = KMeans(n_clusters=k, init="k-means++", n_init=N_INIT, random_state=SEED).fit(km_Z["train"])
    sizes = np.bincount(model.labels_, minlength=k)
    sil = safe_silhouette(km_Z["train"], model.labels_)
    rows.append({"k": k, "silhouette": sil, "inertie": model.inertia_,
                 "davies_bouldin": davies_bouldin_score(km_Z["train"], model.labels_) if not np.isnan(sil) else np.nan,
                 "groupes_formes": int((sizes > 0).sum()), "degenere": bool(np.isnan(sil)),
                 "plus_petit_groupe": int(sizes.min()), "effectifs": "/".join(map(str, sorted(sizes, reverse=True))),
                 "init": "k-means++", "n_init": N_INIT, "random_state": SEED, "donnees": "train standardisé"})
    km_models[k] = model

km_scores = pd.DataFrame(rows)
km_scores.to_csv(RESULTS / "silhouette_scores.csv", index=False)
display(km_scores.drop(columns=["init", "n_init", "random_state", "donnees"]).round(4))

ranking = km_scores.sort_values("silhouette", ascending=False)
K = int(ranking.iloc[0]["k"])
K_SECOND = int(ranking.iloc[1]["k"])
SIL_GAP = float(ranking.iloc[0]["silhouette"] - ranking.iloc[1]["silhouette"])
print(f"k retenu (silhouette maximale parmi les valeurs testées) : {K}")
print(f"Deuxième valeur : k = {K_SECOND}, écart de silhouette = {SIL_GAP:.3f}")
print("Solutions dégénérées :", int(km_scores["degenere"].sum()))

fig, axes = plt.subplots(1, 3, figsize=(13, 3.5))
for ax, col, title in zip(axes, ["silhouette", "inertie", "davies_bouldin"],
                          ["Silhouette (à maximiser)", "Inertie — méthode du coude", "Davies-Bouldin (à minimiser)"]):
    ax.plot(km_scores["k"], km_scores[col], "o-", color="#3a8f85")
    ax.plot(K, km_scores.set_index("k").loc[K, col], "o", color="#d97757", ms=10, label=f"k retenu = {K}")
    ax.set(title=title, xlabel="k")
axes[0].legend()
plt.tight_layout()
plt.savefig(FIGURES / "choix_k.png", bbox_inches="tight")
plt.show()
````

## 4. Amélioration étudiée : initialisation `random` contre `k-means++`

* **Variante standard** — `init="random"` : les k centres initiaux sont k étudiants tirés au hasard. Deux centres peuvent tomber dans la même zone dense et l'algorithme de Lloyd converge alors vers un minimum local.
* **Amélioration** — `init="k-means++"` (Arthur et Vassilvitskii, 2007) : le premier centre est tiré au hasard, chaque centre suivant est tiré avec une probabilité proportionnelle à $D(x)^2$, la distance au carré au centre déjà choisi le plus proche. Les centres de départ sont donc étalés.

Protocole : k fixé à la valeur retenue, mêmes graines pour les deux variantes (42 à 61, dont 42, 43 et 44), même `n_init`. On mesure inertie, silhouette, nombre d'itérations, temps d'ajustement et ARI par rapport à la solution de référence de la section 3. Deux réglages de `n_init` sont testés :

* `n_init=1` — une seule initialisation, ce qui révèle la sensibilité au point de départ ;
* `n_init=20` — le réglage utilisé en pratique (on garde la meilleure inertie sur 20 départs).

**Règle de choix, fixée avant de lire les résultats (critères internes uniquement) :** avec `n_init=1`, retenir la variante qui atteint le plus souvent la meilleure inertie connue, puis la plus faible inertie moyenne, puis le moins d'itérations.

L'avantage théorique de k-means++ grandit avec le nombre de groupes, car il devient plus probable qu'un tirage uniforme place deux centres dans la même zone. On répète donc la même mesure à k = 8, à titre de contrôle complémentaire. La règle de choix ne porte que sur le k retenu.

````python
INIT_SEEDS = list(range(42, 62))
K_CONTROL = 8
km_reference = km_models[K].labels_

rows = []
for k_cmp in [K, K_CONTROL]:
    for n_init in [1, N_INIT]:
        for init in ["random", "k-means++"]:
            for seed in INIT_SEEDS:
                started = perf_counter()
                model = KMeans(n_clusters=k_cmp, init=init, n_init=n_init, random_state=seed).fit(km_Z["train"])
                elapsed = perf_counter() - started
                rows.append({"k": k_cmp, "variante": init, "n_init": n_init, "graine": seed,
                             "inertie": model.inertia_, "silhouette": safe_silhouette(km_Z["train"], model.labels_),
                             "iterations": int(model.n_iter_), "temps_s": elapsed,
                             "ari_vs_reference": adjusted_rand_score(km_models[k_cmp].labels_, model.labels_)})
km_init = pd.DataFrame(rows)
best_by_k = km_init.groupby("k")["inertie"].transform("min")
km_init["atteint_meilleure_inertie"] = km_init["inertie"] <= best_by_k * (1 + 1e-6)
km_init.to_csv(RESULTS / "initialization_comparison.csv", index=False)
best_inertia = km_init.loc[km_init["k"] == K, "inertie"].min()

km_init_summary = km_init.groupby(["k", "n_init", "variante"]).agg(
    inertie_moy=("inertie", "mean"), inertie_std=("inertie", "std"), inertie_max=("inertie", "max"),
    silhouette_moy=("silhouette", "mean"), silhouette_min=("silhouette", "min"),
    iterations_moy=("iterations", "mean"), temps_moy_ms=("temps_s", lambda s: 1000 * s.mean()),
    ari_moy=("ari_vs_reference", "mean"), ari_min=("ari_vs_reference", "min"),
    part_meilleure_inertie=("atteint_meilleure_inertie", "mean"))
display(km_init_summary.round(4))

single = km_init_summary.loc[(K, 1)].sort_values(["part_meilleure_inertie", "inertie_moy", "iterations_moy"],
                                            ascending=[False, True, True])
INIT_CHOICE = single.index[0]
print("Meilleure inertie connue :", round(best_inertia, 3))
print("Variante retenue par la règle :", INIT_CHOICE)
````

## 5. Code personnalisé : K-means réimplémenté avec NumPy

Pour vérifier notre compréhension de l'algorithme et ne pas dépendre d'une boîte noire, on réimplémente K-means avec NumPy :

1. **Initialisation** `random` ou `k-means++` (tirage proportionnel à $D(x)^2$).
2. **Affectation** : chaque étudiant rejoint le centre le plus proche (distance euclidienne au carré).
3. **Mise à jour** : chaque centre devient la moyenne de ses membres. Un groupe vide est réinitialisé sur le point le plus éloigné de son centre.
4. Répéter 2–3 jusqu'à stabilité des centres. Avec `n_init` départs, garder la solution d'inertie minimale $\sum_i \lVert x_i - \mu_{c(i)} \rVert^2$.

On teste d'abord la fonction sur trois nuages de points bien séparés, puis on compare sa solution à celle de scikit-learn sur nos données.

````python
def squared_distances(X, centers):
    """Matrice n x k des distances euclidiennes au carré."""
    return ((X[:, None, :] - centers[None, :, :]) ** 2).sum(axis=2)


def init_centers(X, k, init, rng):
    if init == "random":
        return X[rng.choice(len(X), size=k, replace=False)].copy()
    centers = [X[rng.integers(len(X))]]
    d2 = ((X - centers[0]) ** 2).sum(axis=1)
    for _ in range(1, k):
        probs = d2 / d2.sum() if d2.sum() > 0 else np.full(len(X), 1 / len(X))
        new_center = X[rng.choice(len(X), p=probs)]
        centers.append(new_center)
        d2 = np.minimum(d2, ((X - new_center) ** 2).sum(axis=1))
    return np.array(centers)


def kmeans_numpy(X, k, init="k-means++", n_init=10, max_iter=300, tol=1e-8, seed=42):
    """K-means (Lloyd) avec NumPy. Retourne centres, étiquettes, inertie et nombre d'itérations."""
    X = np.asarray(X, dtype=float)
    rng = np.random.default_rng(seed)
    best = None
    for _ in range(n_init):
        centers = init_centers(X, k, init, rng)
        for iteration in range(1, max_iter + 1):
            d2 = squared_distances(X, centers)
            assignment = d2.argmin(axis=1)
            new_centers = centers.copy()
            for j in range(k):
                members = assignment == j
                if members.any():
                    new_centers[j] = X[members].mean(axis=0)
                else:  # groupe vide : repartir du point le plus mal représenté
                    new_centers[j] = X[d2.min(axis=1).argmax()]
            shift = ((new_centers - centers) ** 2).sum()
            centers = new_centers
            if shift <= tol:
                break
        d2 = squared_distances(X, centers)
        assignment = d2.argmin(axis=1)
        inertia = float(d2[np.arange(len(X)), assignment].sum())
        if best is None or inertia < best["inertia"]:
            best = {"centers": centers, "labels": assignment, "inertia": inertia, "n_iter": iteration}
    return best


# Vérification sur trois nuages séparés : la partition doit être retrouvée exactement
toy_rng = np.random.default_rng(0)
toy_truth = np.repeat([0, 1, 2], 50)
toy_X = np.vstack([toy_rng.normal(center, 0.3, size=(50, 2)) for center in [(0, 0), (5, 5), (0, 6)]])
toy = kmeans_numpy(toy_X, 3, seed=0)
assert adjusted_rand_score(toy_truth, toy["labels"]) == 1.0
assert np.bincount(toy["labels"]).tolist() == [50, 50, 50]
print("Test sur données synthétiques réussi.")

started = perf_counter()
km_custom = kmeans_numpy(km_Z["train"], K, init="k-means++", n_init=N_INIT, seed=SEED)
custom_time = perf_counter() - started
sk_inertia = km_models[K].inertia_
print(f"Inertie NumPy       : {km_custom['inertia']:.3f}  ({custom_time:.2f} s)")
print(f"Inertie scikit-learn: {sk_inertia:.3f}")
print(f"Écart relatif       : {(km_custom['inertia'] - sk_inertia) / sk_inertia:+.2e}")
print(f"ARI entre les deux partitions : {adjusted_rand_score(km_models[K].labels_, km_custom['labels']):.4f}")
````

## 8. Répartition des étiquettes réelles dans chaque groupe

Le modèle étant fixé, on joint maintenant les vraies étiquettes par `row_id`. Pour chaque split, on calcule les effectifs et les pourcentages de `Dropout`, `Enrolled` et `Graduate` dans chaque groupe (les pourcentages sont calculés sur l'effectif du groupe et totalisent 100 %). On compare le taux de décrochage de chaque groupe au taux du **même** split avec le rapport `ratio_dropout` = taux du groupe / taux du split. Un rapport supérieur à 1 signale une sur-représentation des décrocheurs.

Les étiquettes de validation et de test servent seulement à décrire le résultat final ; elles n'interviennent ni dans le choix de k ni dans l'ajustement.

````python
rows = []
for split in SPLITS:
    y = km_y[split]
    split_dropout_pct = 100 * (y == "Dropout").mean()
    counts = pd.crosstab(km_clusters[split], y).reindex(index=range(K), columns=CLASSES, fill_value=0)
    for c in range(K):
        n = int(counts.loc[c].sum())
        row = {"split": split, "cluster": c, "effectif": n, "part_du_split_pct": 100 * n / len(y)}
        row.update({f"n_{cls}": int(counts.loc[c, cls]) for cls in CLASSES})
        row.update({f"pct_{cls}": 100 * counts.loc[c, cls] / n if n else np.nan for cls in CLASSES})
        row["taux_dropout_split_pct"] = split_dropout_pct
        row["ratio_dropout"] = row["pct_Dropout"] / split_dropout_pct if n else np.nan
        rows.append(row)
km_profiles = pd.DataFrame(rows)
pct_cols = [f"pct_{cls}" for cls in CLASSES]
assert np.allclose(km_profiles[pct_cols].sum(axis=1), 100)
km_profiles.to_csv(RESULTS / "cluster_profiles.csv", index=False)
display(km_profiles.round(1))

rows = []
for split in SPLITS:
    y, cl = km_y[split], km_clusters[split]
    counts = pd.crosstab(cl, y)
    rows.append({"split": split, "purete": counts.max(axis=1).sum() / len(y),
                 "reference_classe_majoritaire": y.value_counts(normalize=True).max(),
                 "ARI": adjusted_rand_score(y, cl), "NMI": normalized_mutual_info_score(y, cl)})
km_agreement = pd.DataFrame(rows).set_index("split")
display(km_agreement.round(3))
````

## 9. Lecture supervisée a posteriori (fonction `map_labels` du cours)

Comme dans le support du cours (*KMeans — groupement des caractères*), on associe à chaque groupe la classe majoritaire de ses membres sur train, puis on applique ce mapping aux groupes prédits pour validation et test. Ce n'est pas un classifieur optimisé : cela mesure seulement à quel point les groupes « naturels » recouvrent les classes. Le résultat ne sert pas à choisir k.

````python
def map_labels(km_labels, y):
    """Reprise de la fonction du cours : classe majoritaire de chaque groupe."""
    mapping_ = {}
    for i in np.unique(km_labels):
        mapping_[i] = pd.Series(y)[km_labels == i].value_counts().reindex(CLASSES).idxmax()
    return mapping_


km_map = map_labels(km_clusters["train"], km_y["train"].to_numpy())
print("Groupe -> classe majoritaire :", km_map)
rows = []
for split in SPLITS:
    pred = pd.Series(km_clusters[split]).map(km_map).to_numpy()
    rows.append({"split": split, "accuracy": (pred == km_y[split].to_numpy()).mean(),
                 "f1_macro": f1_score(km_y[split], pred, labels=CLASSES, average="macro", zero_division=0),
                 "classes_jamais_predites": ", ".join(sorted(set(CLASSES) - set(pred))) or "aucune"})
km_mapping_scores = pd.DataFrame(rows).set_index("split")
display(km_mapping_scores.round(3))

task2_metrics = ROOT / "results" / "task2" / "test_metrics.csv"
if task2_metrics.exists():
    print("Pour comparaison, F1 macro test de la tâche 2 :")
    display(pd.read_csv(task2_metrics)[["model", "macro_f1"]].round(3))
````
